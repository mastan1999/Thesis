```python
import pandas as pd
import numpy as np

# تنظیمات نمایش ستون‌ها برای دیدن تمام اطلاعات در خروجی
pd.set_option('display.max_columns', None)

# ------------------------------------------------------------------
# --- مرحله ۱: خواندن فایل‌های خام (مسیردهی نهایی و صحیح: '../') ---
# ------------------------------------------------------------------
# توجه: مسیردهی '..' به معنای یک پوشه به عقب برگشتن از notebooks/ به Thesis/ است.
try:
    # --- فایل‌های ETF اصلی ---
    df_bito = pd.read_csv('../BITO.csv')
    df_bft = pd.read_csv('../BTF.csv')
    
    # --- فایل‌های بسته شدن قیمت و نتایج قبلی (برای تحلیل) ---
    df_bito_close = pd.read_csv('../BITO_Closing_Prices.csv')
    df_gbtc_close = pd.read_csv('../GBTC_Closing_Prices.csv')
    df_gbtc_result = pd.read_csv('../gbtc_premium_discount_result.csv')

    print("✅ فایل‌ها با موفقیت بارگذاری شدند. آماده بازرسی!")
    
except FileNotFoundError as e:
    print(f"❌ خطا: فایل‌ها در مسیر مشخص شده یافت نشدند. خطا: {e}")
    # اگر این خطا مجدداً رخ داد، لطفاً نام فایل یا محل آن‌ها را بررسی کنید.
    # مثلاً ممکن است نام یکی از فایل‌ها دقیقاً GBTC_Closing_Prices.csv نباشد.
    
# ----------------------------------------------------
# --- مرحله ۲: تابع کمکی برای بازرسی سریع ---
# ----------------------------------------------------

def inspect_data(df, name):
    """تابع برای نمایش اطلاعات کلیدی DataFrame"""
    print("\n" + "="*50)
    print(f"         بازرسی اولیه: {name}")
    print("="*50)

    # بررسی ساختار (انواع داده و مقادیر غیر تهی)
    print(f"\n--- {name} Info (انواع داده و شمارش ردیف‌ها) ---")
    df.info()

    # بررسی مقادیر گمشده (Missing Values)
    print(f"\n--- {name} Missing Values (مقادیر گمشده) ---")
    print(df.isnull().sum())

    # بررسی ۵ سطر اول
    print(f"\n--- {name} Head (نمونه اولیه) ---")
    print(df.head())

# ----------------------------------------------------
# --- مرحله ۳: اجرای بازرسی بر روی تمام داده‌ها ---
# ----------------------------------------------------

inspect_data(df_bito, 'BITO_RAW')
inspect_data(df_bft, 'BTF_RAW')
inspect_data(df_bito_close, 'BITO_Closing_Prices')
inspect_data(df_gbtc_close, 'GBTC_Closing_Prices')
inspect_data(df_gbtc_result, 'GBTC_Premium_Discount_Result')
```
