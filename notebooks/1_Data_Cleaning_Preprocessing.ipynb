```python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt   # data visualization


# تنظیمات نمایش ستون‌ها برای دیدن تمام اطلاعات در خروجی
pd.set_option('display.max_columns', None)

# ------------------------------------------------------------------
# --- مرحله ۱: خواندن فایل‌های مورد نیاز (فقط BITO و BTF) ---
# ------------------------------------------------------------------
try:
    # مسیر صحیح: یک پوشه به عقب برگرد ('..') و فایل را در ریشه Thesis بخوان.
    df_bito = pd.read_csv('../BITO.csv')
    df_bft = pd.read_csv('../BTF.csv')
    
    print("✅ فایل‌های BITO و BTF با موفقیت بارگذاری شدند. آماده بازرسی!")
    
except FileNotFoundError as e:
    # اگر باز هم خطا داد، نام فایل یا مسیر را چک کنید.
    print(f"❌ خطا: فایل‌ها در مسیر مشخص شده یافت نشدند. خطا: {e}")
    raise # اجرای کد متوقف می شود تا فایل را اصلاح کنید
    
# ----------------------------------------------------
# --- مرحله ۲: تابع کمکی برای بازرسی سریع ---
# ----------------------------------------------------

def inspect_data(df, name):
    """تابع برای نمایش اطلاعات کلیدی DataFrame: info, isnull.sum, head"""
    print("\n" + "="*50)
    print(f"         بازرسی اولیه: {name}")
    print("="*50)

    # بررسی ساختار (انواع داده و مقادیر غیر تهی)
    print(f"\n--- {name} Info (انواع داده و شمارش ردیف‌ها) ---")
    # اطمینان حاصل می کنیم که df تعریف شده باشد
    if df is not None:
        df.info()

        # بررسی مقادیر گمشده (Missing Values)
        print(f"\n--- {name} Missing Values (مقادیر گمشده) ---")
        print(df.isnull().sum())

        # بررسی ۵ سطر اول
        print(f"\n--- {name} Head (نمونه اولیه) ---")
        print(df.head())
    else:
        print(f"DataFrame {name} تعریف نشده است.")

# ----------------------------------------------------
# --- مرحله ۳: اجرای بازرسی بر روی دو داده مورد نظر ---
# ----------------------------------------------------

# مطمئن شوید که تابع فقط در صورت بارگذاری موفقیت‌آمیز اجرا شود
if 'df_bito' in locals() and 'df_bft' in locals():
    inspect_data(df_bito, 'BITO_RAW')
    inspect_data(df_bft, 'BTF_RAW')
else:
    print("\n⚠️ بازرسی انجام نشد. ابتدا مشکل FileNotFoundError را حل کنید.")
```
